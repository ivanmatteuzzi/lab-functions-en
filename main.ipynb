```python
def greatest(*numbers):
    biggest = numbers[0]

    for number in numbers:
        if number > biggest:
            biggest = number

    return biggest
          
```

## 5. Now combine those two ideas and write a function that receives a list and either "+" or "*" and outputs acordingly

```python
def calculate(numbers, operation):
    if operation == "+":
        result = 0
        for number in numbers:
            result += number
        return result

    elif operation == "*":
        result = 1
        for number in numbers:
            result *= number
        return result
```

```python
print(calculate([2, 3, 4], "+"))
print(calculate([2, 3, 4], "*"))
```

## 6. Write a function that returns the factorial of a number.

```python
def factorial(n):
    result = 1

    for i in range(1, n + 1):
        result = result * i
    return result
```

```python
print(factorial(6))
```

## 8. Write a function that returns the mode of a list, i.e.: the element that appears the most times.
`NOTE: You should not use count... 🧐`

```python
def mode_counter(arr):
    frequency = {}

    for item in arr:
        if item not in frequency:
            frequency[item] = 1

        else:
            frequency[item] += 1

    mode = None
    highest = 0

    for item in frequency:
        if frequency[item] > highest:
            highest = frequency[item]
            mode = item

    return mode
```

```python
print(mode_counter([1, 2, 2, 3, 3, 3, 4]))
```

## 9. Write a function that calculates the standard deviation of a list.
`NOTE: Do not use any libraries or already built functions. 😉`

```python
times = [54.2, 54.8, 55.1, 54.5, 55.4]

def standard_deviation(times):
    total = 0

    for time in times:
        total += time

    mean = total / len(times)

    squared_differences = 0

    for time in times:
        difference = time - mean
        squared_differences += difference ** 2

    return (squared_differences / len(times)) ** 0.5
```

```python
print(standard_deviation(times))
```

## 12. Write a function to check if a given password is strong (at least 8 characters, at least one lower case, at least one upper case, at least one number and at least one special character). It should output True if strong and False if not.

```python
def check_pass(password):
    has_lower = False
    has_upper = False
    has_number = False
    has_special = False

    for char in password:
        if char.islower():
            has_lower = True
        elif char.isupper():
            has_upper = True
        elif char.isdigit():
            has_number = True
        else:
            has_special = True

    return len(password) >= 8 and has_lower and has_upper and has_number and has_special


```

```python
print(check_pass("Nuotatore99!"))
```
